# Social-LSTM

```python
from tensorboard.backend.event_processing import event_accumulator
import pandas as pd

# 加载events文件
ea = event_accumulator.EventAccumulator('runs_social_lstm')
ea.Reload()

all_data = {}
for tag in ea.Tags()['scalars']:
    data = ea.Scalars(tag)
    all_data[tag] = {item.step: item.value for item in data}

df = pd.DataFrame(all_data)
df.index.name = 'step'
```

```python
df = df[(df.index % 10) ==0]
```
